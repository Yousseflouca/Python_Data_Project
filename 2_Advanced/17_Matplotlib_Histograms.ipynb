```python
import pandas as pd
import matplotlib.pyplot as plt
import ast

df = pd.read_csv('data_jobs.csv')

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DA_US = df[(df['job_title_short'] == 'Data Analyst') & (df['job_country'] == 'United States')].copy()
```

```python
df_DA_US['salary_year_avg'].plot(kind='hist', bins=40, edgecolor='black')
plt.xlim(0, 250000)

ax=plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${int(x/1000)}K'))

plt.title('Distribution of United States Data Analyst Yearly Salaries')
plt.xlabel('Yearly Salary')
plt.ylabel('Number of Jobs')
```

```python
df_DS = df[df['job_title_short'] == 'Data Scientist']
df_DS['salary_hour_avg'].plot(kind='hist', bins=40, color='blue', edgecolor='black')
plt.xlabel('Hourly Salary')
plt.ylabel('Number of Jobs')
plt.title('Distribution of Data Scientist Hourly Salaries')
plt.xlim(0,150)
plt.tight_layout()
plt.show()
```

```python
df_skills = df[df['job_skills'].notna()]
df_skills['skill_count'] = df_skills['job_skills'].apply(lambda x: len(x))

df_skills['skill_count'].plot(kind='hist', bins=55, color='magenta', edgecolor='black')
plt.xlabel('Number of Skills')
plt.ylabel('Number of Job Postings')
plt.title('Distribution of Job Postings by Number of Skills Listed')
plt.xlim(0, 25)
plt.tight_layout()
plt.show()
```

```python
df_DE = df[df['job_title_short'] == 'Data Engineer'].copy()

mean_salary = df_DE['salary_year_avg'].mean()
median_salary = df_DE['salary_year_avg'].median()

df_DE['salary_year_avg'].plot(kind='hist', bins=50, color='red', edgecolor='black')
mean_line = plt.axvline(
    x=mean_salary,
    color='blue',
    linestyle='dashed',
    label=f'Mean: ${mean_salary:,.0f}'
)
median_line = plt.axvline(
    x=median_salary,
    color='green',
    linestyle='--',
    label=f'Median: ${median_salary:,.0f}'
)
plt.legend(handles=[mean_line, median_line])
plt.xlim(0,300000)
plt.xlabel('Yearly Salary ($)')
plt.ylabel('Number of Jobs')
plt.title('Data Engineer Salary Distribution')
plt.show()
```
