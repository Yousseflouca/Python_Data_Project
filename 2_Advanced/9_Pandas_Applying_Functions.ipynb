```python
df_salary = df[pd.notna(df['salary_year_avg'])].copy()
df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(lambda salary: salary*1.03)

df_salary[['salary_year_avg', 'salary_year_inflated']]
```

```python
def projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return row['salary_year_avg'] * 1.05
    else:
        return row['salary_year_avg'] * 1.03

df_salary['salary_year_inflated'] = df_salary.apply(projected_salary, axis=1)

df_salary[['job_title_short', 'salary_year_avg', 'salary_year_inflated']].head(20)
```

```python
df_date = df.copy()
```

```python
def date_str(date):
    year = date.strftime('%Y')
    month = date.strftime('%m')
    day = date.strftime('%d')
    return year + '-' + month + '-' + day

df_date['job_posted_date_str'] = df_date['job_posted_date'].apply(date_str)

df_date[['job_posted_date', 'job_posted_date_str']]
```

```python
df_date['job_posted_date_str'] = df_date['job_posted_date'].dt.strftime('%Y-%m-%d')

df_date[['job_posted_date', 'job_posted_date_str']]
```

```python
import datetime

df_date['days_since_posted'] = df_date['job_posted_date'].apply(lambda x: (datetime.datetime.now() - x).days)
```

```python
df_date[['job_posted_date', 'days_since_posted']]
```

```python
df_filtered = df[pd.notna(df['salary_year_avg'])].copy()

df_filtered['salary_category'] = df_filtered['salary_year_avg'].apply(lambda salary: 'Low' if salary < 60_000 else 'Medium' if salary < 100_000 else 'High')

df_filtered[['salary_year_avg', 'salary_category']]
```
