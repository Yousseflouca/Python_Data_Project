```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data_jobs.csv')
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.loc[:10, 'salary_year_avg': 'salary_hour_avg']
```

```python
median_salary_year = df['salary_year_avg'].median()
```

```python
median_salary_hour = df['salary_hour_avg'].median()
```

```python
df_filled = df

df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df_filled.loc[:10, 'salary_year_avg': 'salary_hour_avg']
```

```python
df_unique = df_filled

df_unique = df_unique.drop_duplicates()

print('Length of original df:       ', len(df_filled))
print('Length of drop duplicates df:', len(df_unique))
print('Rows Dropped:                ', len(df_filled)-len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])

print('Length of original df:       ', len(df_filled))
print('Length of drop duplicates df:', len(df_unique))
print('Rows Dropped:                ', len(df_filled)-len(df_unique))
```

```python
df_remove_missing = df

df_remove_missing = df_remove_missing.dropna(subset=['salary_year_avg'])

print('Length of original df:       ', len(df))
print('Length of drop missing df:   ', len(df_remove_missing))
print('Rows Dropped:                ', len(df)-len(df_remove_missing))
```

```python
df_remove_duplicates = df

df_remove_duplicates = df_remove_duplicates.drop_duplicates(subset='job_location')

print('Length of original df:       ', len(df))
print('Length of drop duplicated df:', len(df_remove_duplicates))
print('Rows Dropped:                ', len(df)-len(df_remove_duplicates))
```
